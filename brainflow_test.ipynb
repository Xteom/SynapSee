## Setup Open BCI helmet

OpenBCI GUI, click on `Cython (live)` -> `Serial (from Dongle)`, make sure to have everything as the image below:

![OpenBCI_GUI-1.png](docs_images/OpenBCI_GUI-1.png)

click on `Auto-Connect`

To see the signal in real time click `Start Data Stream`

![OpenBCI_GUI-2.png](docs_images/OpenBCI_GUI-2.png)

To check the signal of the electrodes select `Cython Signal` in one of the cuadritos (alguien cambie esto)

![OpenBCI_GUI-3.png](docs_images/OpenBCI_GUI-3.png)


Click on `Check All Channels`. When the electrode resistance indicator turns green, the signal is within the desired threshold.

![OpenBCI_GUI-4.png](docs_images/OpenBCI_GUI-4.png)


Once all (or most) electrodes show green, close the GUI and proceed with your experiment. If you leave a stream running in the GUI may result in port issues.
